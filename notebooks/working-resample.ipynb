# Resample the full csv into smaller datasets.
## Using random resampling

```python
import pandas as pd

CSV_PATH = "data/SGJobData.csv"
RANDOM_SEED = 42
```

```python
df = pd.read_csv(CSV_PATH)

print("Original dataset shape:", df.shape)
print("Original columns:")
print(df.columns.tolist())
```

```python
SAMPLE_SIZE = 35_000
New_CSV_Description = f"data/SGJobData_{SAMPLE_SIZE}_random.csv"

# create the smaller dataset ussing random sampling
sample_dataset = df.sample(n=SAMPLE_SIZE, random_state=RANDOM_SEED)

# write the smaller dataset to a new CSV file
sample_dataset.to_csv(New_CSV_Description, index=False)
```
